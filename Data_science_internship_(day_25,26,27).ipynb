```python
import numpy as np
import pandas as pd
df=pd.read_csv('apples_and_oranges.csv')
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.describe()
```

```python
x=df[['Weight','Size']]
y=df.Class
```

```python
df.isnull().sum()
```

```python
y
```

```python
from sklearn.tree import DecisionTreeClassifier
clf=DecisionTreeClassifier()
clf.fit(x,y)
```

```python
ypred=clf.predict(x)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
confusion_matrix(y,ypred)
```

```python
accuracy_score(y,ypred)
```

```python
from sklearn.tree import plot_tree
plot_tree(clf,feature_names=['Weight','Size'],class_names=['orange','apple'])
```

3D plots and wordclouds

```python
from wordcloud import WordCloud,STOPWORDS,ImageColorGenerator
```

```python
text="Coke,Pepsi,read,write,trip,coke,data,visualization"
```

```python
sw=set(STOPWORDS)
sw
```

```python
sw.update(['data','visualization','said'])
sw
```

```python
cloud=WordCloud(background_color="White",stopwords=sw)
wordcloud=cloud.generate(text)
plt.imshow(wordcloud)
plt.axis("off")
plt.show()
```
