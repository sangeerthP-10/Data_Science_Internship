```python
df=df.iloc[:,1:]
df
```

```python
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
cols=['driveway','recroom','fullbase','gashw','airco','prefarea']

for f in cols:
  df[f]=le.fit_transform(df[f])
```

```python
df.head()
```

Simple Linear Regression

```python
df.corr()
```

```python
x=df[['lotsize']]
y=df[['price']]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression
LR=LinearRegression()
LR.fit(x_train,y_train)
```

```python
LR.coef_
```

```python
LR.intercept_
```

```python
y_pred=LR.predict(x_test)
y_pred
```

```python
from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```

Prediction for house price for given lot size

```python
LR.predict([[5000]])
```

Multiple Linear Regression

```python
x1=df[['lotsize','bathrms']]
```

```python
x1_train,x1_test,y1_train,y1_test=train_test_split(x1,y,test_size=0.2,random_state=2)
MLR=LinearRegression()
MLR.fit(x1_train,y1_train)
```
