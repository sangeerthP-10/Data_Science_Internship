```python


import pandas as pd
import numpy as mp
```

```python
df=pd.read_csv("Iris_1.csv")
df
```

```python
df.info()
```

```python

df1=df.copy(deep=True)
```

```python
df1.isna().sum()
```

```python
median1=df['SepalLengthCm'].median()
df1['SepalLengthCm'] = df1['SepalLengthCm'].fillna(median1)
df1.isna().sum()
```

```python
median2=df['SepalWidthCm'].median()
df1['SepalWidthCm'] = df1['SepalWidthCm'].fillna(median2)
df1.isna().sum()
```

```python
median3=df['PetalLengthCm'].median()
df1['PetalLengthCm'] = df1['PetalLengthCm'].fillna(median3)
df1.isna().sum()
```

```python
median4=df['PetalWidthCm'].median()
df1['PetalWidthCm'] = df1['PetalWidthCm'].fillna(median4)
df1.isna().sum()
```
