```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('apples_and_oranges.csv')
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
from sklearn.preprocessing import LabelEncoder
en=LabelEncoder()
df['Class']=en.fit_transform(df['Class'])
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
x=df[['Weight','Size']]
y=df['Class']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

Linear SVM

```python
from sklearn.svm import LinearSVC
classifier=LinearSVC()
classifier.fit(x_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
y_pred=classifier.predict(x_test)
accuracy_score(y_test,y_pred)
```

confusion matrix

```python
from sklearn.metrics import confusion_matrix,classification_report
confusion_matrix(y_test,y_pred)
```

```python
classification_report(y_test,y_pred)
```

Non linear SVM

```python
df_iris=pd.read_csv('Iris.csv')
df_iris.head()
```

```python
from sklearn.svm import LinearSVC,SVC
X=df_iris.iloc[:,0:5]
Y=df_iris['Species']
lin_SVC=LinearSVC().fit(X,Y)
svc_lin=SVC(kernel='linear').fit(X,Y)#linear decison boundary for seperating data

svc_rbf=SVC(kernel='rbf').fit(X,Y)#non linear boundary

svc_poly1=SVC(kernel='poly',degree=4).fit(X,Y)


svc_poly2=SVC(kernel='poly',degree=4).fit(X,Y)
```

```python
lin_SVC.score(X,Y)
```

```python
svc_lin.score(X,Y)
```

```python
svc_rbf.score(X,Y)
```

```python
svc_poly1.score(X,Y)
```

```python
svc_poly2.score(X,Y)
```
